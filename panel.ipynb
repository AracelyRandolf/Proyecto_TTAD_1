```python
import json

with open("data.json", "r", encoding="utf-8") as file:
    data = json.load(file)

ventas = data["ventas"]
```

```python
# Productos con mas ingresos
 
ingresos_producto = {}

for venta in ventas:

    producto = venta["producto"]

    ingresos = venta["cantidad"] * venta["precio_unitario"]

    ingresos_producto[producto] = (
        ingresos_producto.get(producto, 0)
        + ingresos
    )

producto_mas_ingresos = max(
    ingresos_producto,
    key=ingresos_producto.get
)

print("Producto con más ingresos:")
print(producto_mas_ingresos, "-", ingresos_producto[producto_mas_ingresos], "€")
```

```python
# Filtrar ventas por vendedor

vendedor = "Luis"

for venta in ventas:

    if venta["vendedor"] == vendedor:
        print(venta)
```

```python
# Ticket medio
  # Dinero total/ Número de ventas

total = 0

for venta in ventas:

    total = total + (
        venta["cantidad"] * venta["precio_unitario"]
    )

ticket_medio = total / len(ventas)

print("Ticket medio:", ticket_medio, "€")
```
